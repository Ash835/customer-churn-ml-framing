# Rule-Based Customer Churn Baseline

## Objective
Implement and evaluate a transparent rule-based
baseline for identifying potentially at-risk customers.

## Rule
Flag customers when:
- last_login_days >= 14
- support_tickets >= 3

Both conditions must be satisfied.

```python
import sys
sys.path.append("../src")

import pandas as pd
from baseline_rules import add_risk_predictions

df = pd.read_csv("../data/customer-churn-training.csv")

print("Dataset loaded!")
display(df.head())
```

```python
results = add_risk_predictions(df)

display(
    results[
        [
            "customer_id",
            "last_login_days",
            "support_tickets",
            "churned",
            "predicted_risk",
            "predicted_churn"
        ]
    ]
)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

y_true = results["churned"]
y_pred = results["predicted_churn"]

print("Baseline Evaluation")
print("-------------------")

print("Accuracy:", accuracy_score(y_true, y_pred))
print("Precision:", precision_score(y_true, y_pred, zero_division=0))
print("Recall:", recall_score(y_true, y_pred, zero_division=0))
print("F1 Score:", f1_score(y_true, y_pred, zero_division=0))

print("\nConfusion Matrix:")
print(confusion_matrix(y_true, y_pred, labels=[0, 1]))

print("\nClassification Report:")
print(classification_report(y_true, y_pred, zero_division=0))
```

## Baseline Results and Interpretation

The rule-based baseline achieved the following results on the supplied 12-record dataset:

- Accuracy: 91.67%
- Precision: 100%
- Recall: 80%
- F1-score: 88.89%

The confusion matrix showed 7 true negatives, 4 true positives, 1 false negative, and 0 false positives.

The baseline successfully identified four out of five customers labelled as churned, while missing one churned customer.

The absence of false positives suggests that the selected rule was conservative on this small dataset. However, the missed churn case demonstrates that the rule does not identify every churned customer.

These results are descriptive of the 12 supplied records only. Since the rule was evaluated on the same dataset used during exploration, the metrics must not be interpreted as evidence of generalization or production readiness.

A larger, representative dataset and independent evaluation are necessary before drawing conclusions about predictive performance.
